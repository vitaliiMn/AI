## Семинар 3.4 PCA

# Генерация данных

```python
import numpy as np
import matplotlib.pylab as plt 

N = 1000
W = np.array([[5,2], [2,4]])
m = np.array([2,2])
X = np.random.randn(N,2)@W + m

fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111)
plt.scatter(X[:,0], X[:,1])
ax.set_xlabel('x')
ax.set_ylabel('y')
```

## Выборочная ковариация

```python
R = np.cov(X.T)
```

## Главные компоненты

```python
evl, ev = np.linalg.eig(R)
```

```python
ev
```

```python
pb=1
pl=1
#plt.figure(figsize=(8, 8))
fig = plt.figure(figsize=(7,7))
ax = fig.add_subplot(111)
#fig, ax = plt.subplots()
ax.set_aspect('equal', adjustable='box')
ax.scatter(X[:,0], X[:,1])
evl = evl
t = np.linspace(0, 2*np.pi, 100)
ax.arrow(m[0],m[1],ev[0,0]*evl[0],ev[1,0]*evl[0],color='r',lw=3,head_width=pb,head_length=pl)
ax.arrow(m[0],m[1],ev[0,1]*evl[1],ev[1,1]*evl[1],color='r',lw=3,head_width=pb,head_length=pl)
#Найдем матрицу H преобразования в новый базис СВ
e = np.array([[1,0],[0,1]])
H = e@ev
x = evl[0]*np.cos(t)
y = evl[1]*np.sin(t)
v = np.vstack([x,y])
#Преобразование эдипса в новый базис
v = H@v
print(v.shape)
ax.plot(v[0,:],v[1,:])
ax.set_xlabel('x')
ax.set_ylabel('y')
#fig.savefig('../../slides/4/pics/pca_1.png')
```
